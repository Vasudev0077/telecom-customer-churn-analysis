```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df
```

```python
sns.barplot(x=df['ContractType'], y=df['MonthlyCharges'], hue=df['Churn'])
plt.title('Average Monthly Charges Across Contract Tiers and Churn Status')
plt.xlabel('Contract Type Tier')
plt.ylabel('Mean Monthly Recurring Charges ($)')
plt.show()
```

#### Key Analytics & Takeaways:
* **Observation:** Compares the financial impact of monthly recurring fees against contract terms.
* **Critical Finding:** High recurring monthly fees are strongly associated with cancellation behavior among short-term Month-to-Month users, whereas long-term contract users remain insensitive to price changes.
* **Strategic Action:** Introduce early pricing discounts on Month-to-Month plans or offer proactive 1-year transition incentives to high-paying accounts.


```python
sns.scatterplot(x=df['TenureMonths'], y=df['TotalCharges'], hue=df['Churn'], size=df['MonthlyCharges'])
plt.title('Multi-Variable Interaction: Tenure vs. Lifetime Value Growth')
plt.xlabel('Customer Tenure (Months)')
plt.ylabel('Total Charges Accumulated ($)')
plt.show()
```

#### Key Analytics & Takeaways:
* **Observation:** Explores multi-dimensional correlations between account age, periodic spending, and cumulative revenue.
* **Critical Finding:** Churned accounts are highly concentrated in the low-tenure, low-value quadrant, visibly mapping the strict mathematical growth path of customer acquisition economics.
* **Data Science Note:** The structural distribution perfectly mirrors the deterministic formula $TotalCharges = Tenure \times MonthlyCharges$.


```python
sns.lineplot(x=df['TenureMonths'], y=df['TotalCharges'], hue=df['Churn'])
plt.title('Longitudinal Spending Trajectories over Customer Lifespan')
plt.xlabel('Customer Tenure (Months)')
plt.ylabel('Total Charges Accumulated ($)')
plt.show()
```

#### Key Analytics & Takeaways:
* **Observation:** Customer lifetime spend tracks linearly over time for both groups.
* **Critical Finding:** Revenue volatility increases significantly after 40 months of tenure.
* **Data Engineering Note:** This variance indicates diverging long-term customer behavioral profiles.


```python
sns.heatmap(pd.crosstab(df["PaymentMethod"], df["ContractType"], 
                        values=df["Churn"].map({"Yes": 1, "No": 0}),
                        aggfunc="mean"), annot=True, fmt=".1%")
#annot=True and fmt='.1%' forces Seaborn to display the actual percentages clearly inside 
#each block. values=... and aggfunc='mean' forces the table to look at the Churn column and
#calculate the average (churn rate) instead of just counting total headers.
plt.title('Categorical Churn Risk Matrix by Billing and Contract Profile')
plt.xlabel('Subscription Contract Type')
plt.ylabel('Customer Payment Method')
plt.show()
```

#### Key Analytics & Takeaways:
* **Observation:** Visualizes cross-categorical cancellation risks across specific subscription plans and billing platforms.
* **Critical Finding:** Customers combining Month-to-Month contracts with Bank Transfers or Unknown payment profiles represent the platform's highest risk clusters, with churn rates approaching 50%.
* **Business Recommendation:** Transition manual payment workflows into automated clearing house (ACH) autopay models to mitigate transactional friction.


```python
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='Churn', palette='Set2')
plt.title('Baseline Distribution of Churn (Class Balance)')
plt.ylabel('Number of Customers')
plt.show()
```

#### Key Analytics & Takeaways:
* **Observation:** Measures the true distribution scale of active vs. canceled accounts.
* **Critical Finding:** The dataset exhibits a notable class imbalance toward retained users.
* **Machine Learning Prep:** Standard accuracy metrics will be highly misleading; downstream predictive pipelines must utilize SMOTE oversampling or optimize for Precision, Recall, and F1-Score instead.


```python
plt.figure(figsize=(10, 5))
sns.kdeplot(data=df, x='TenureMonths', hue='Churn', fill=True, common_norm=False, palette='crest', alpha=0.5)
plt.title('Customer Retention Danger Zone across Tenure Months')
plt.xlabel('Tenure (Months)')
plt.show()
```

#### Key Analytics & Takeaways:
* **Observation:** Tracks continuous customer lifespans using overlapping probability density estimations.
* **Critical Finding:** An overlapping distribution analysis isolates a high-probability "danger zone" for customer cancellations within the first 12 months of service onboarding.
* **Operations Value:** Focus marketing intervention, customer support check-ins, and user experience onboarding specifically around months 3, 6, and 9.


```python
plt.figure(figsize=(7, 4))
sns.countplot(data=df, x='SeniorCitizen', hue='Churn', palette='coolwarm')
plt.title('Churn Disparity Between Standard Users vs. Senior Citizens')
plt.xticks([0, 1], ['Non-Senior (0)', 'Senior Citizen (1)'])
plt.show()
```

#### Key Analytics & Takeaways:
* **Observation:** Segments account longevity metrics directly against categorical customer age indicators.
* **Critical Finding:** Senior citizens experience a significantly higher rate of subscription churn than non-seniors.
* **UI/UX Strategy:** This dropoff highlights a vulnerable user cohort requiring accessibility improvements, specialized customer service lines, or simplified digital payment paths.
